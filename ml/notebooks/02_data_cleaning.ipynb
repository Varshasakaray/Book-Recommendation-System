## Date Cleaning

Next, I'll clean and standardize any date columns. This usually involves converting them to datetime objects for easier manipulation.

```python
# Identify potential date columns
date_columns = ['date_added', 'date_updated', 'read_at', 'started_at']

for col in date_columns:
    if col in df.columns:
        print(f"Converting column '{col}' to datetime...")
        # Use errors='coerce' to turn unparseable dates into NaT (Not a Time)
        df[col] = pd.to_datetime(df[col], errors='coerce')
    else:
        print(f"Column '{col}' not found.")

print('\nData types after date conversion:')
display(df.info())
```

## Invalid Date Check

After converting to datetime objects, it's important to check for any invalid dates (represented as `NaT`) and decide how to handle them. For now, I'll just show the count of `NaT` values.

```python
for col in date_columns:
    if col in df.columns:
        invalid_dates_count = df[col].isna().sum()
        if invalid_dates_count > 0:
            print(f"Column '{col}': {invalid_dates_count} invalid date(s) found (NaT).")
        else:
            print(f"Column '{col}': No invalid dates found.")

# Display rows with invalid dates for inspection
# For simplicity, let's just display a sample of rows where any of the identified date columns is NaT
# You might want to refine this to show specific columns or handle them differently
problematic_rows = df[df[date_columns].isna().any(axis=1)]
if not problematic_rows.empty:
    print('\nSample of rows with invalid dates:')
    display(problematic_rows.head())
else:
    print('\nNo rows with invalid dates across the identified date columns.')
```

## Page Cleaning

Now, let's address the `num_pages` column. This typically involves checking for missing values, incorrect data types, and any outliers that might be present.

```python
print(f"Info for 'num_pages' column before cleaning:")
display(df['num_pages'].describe())
print(f"\nNumber of missing values in 'num_pages': {df['num_pages'].isnull().sum()}")
print(f"\nUnique values in 'num_pages' (first 20 if many):\n{df['num_pages'].value_counts(dropna=False).head(20)}")

# Convert 'num_pages' to numeric, coercing errors
df['num_pages'] = pd.to_numeric(df['num_pages'], errors='coerce')

# Handle missing values: fill with median or mean, or drop rows. For now, let's fill with the median.
# Consider the context before deciding on the best strategy for NaNs.
median_pages = df['num_pages'].median()
print(f"\nFilling missing 'num_pages' with median value: {median_pages}")
df['num_pages'].fillna(median_pages, inplace=True)

# Ensure 'num_pages' are integers (after handling potential float conversions from median/mean)
df['num_pages'] = df['num_pages'].astype(int)

print(f"\nInfo for 'num_pages' column after cleaning:")
display(df['num_pages'].describe())
print(f"\nNumber of missing values in 'num_pages' after cleaning: {df['num_pages'].isnull().sum()}")
```

## Genre String Parsing

The `genres` column appears to contain string representations of lists. I will parse these strings to actual Python lists for easier manipulation and analysis.

```python
import ast
import re # Import the re module for regular expressions

print(f"Original type of 'genres' column: {df['genres'].dtype}")
print(f"Sample of original 'genres' data:\n{df['genres'].head()}")

# Function to parse genre strings using regex
def parse_genres(genre_str):
    if pd.isna(genre_str) or not isinstance(genre_str, str):
        return []
    # Use regex to find all strings enclosed in single quotes
    tags = re.findall(r"'([^']*)'", genre_str)
    return tags

df['genres_list'] = df['genres'].apply(parse_genres)

print(f"\nNew 'genres_list' column created. Sample of parsed 'genres_list' data:\n{df['genres_list'].head()}")
print(f"Type of 'genres_list' column: {df['genres_list'].dtype}")

# df = df.drop('genres', axis=1)
```

## Shelf Tag List

From the `genres_list`, I will now identify and consolidate individual shelf tags. This will involve iterating through the lists and collecting all unique tags.

```python
from collections import defaultdict

# Assuming 'genres_list' contains lists of genres
# If 'genres_list' is not yet created, the previous 'Genre String Parsing' cell needs to be run first.

all_shelf_tags = defaultdict(int)

for index, genre_list in df['genres_list'].items():
    if isinstance(genre_list, list):
        for tag in genre_list:
            all_shelf_tags[tag] += 1

print(f"Total unique shelf tags found: {len(all_shelf_tags)}")
print("Top 20 most frequent shelf tags:")
for tag, count in sorted(all_shelf_tags.items(), key=lambda item: item[1], reverse=True)[:20]:
    print(f"- {tag}: {count}")
```

## Remove Non-Genre Tags

Many tags in the `genres_list` might not represent actual genres but rather reader statuses or organizational labels (e.g., 'to-read', 'currently-reading'). This step aims to identify and remove these non-genre tags to focus on actual genre classifications.

```python
from itertools import chain

# Define a list of common non-genre tags to be removed
# This list can be expanded based on further data exploration
non_genre_tags = [
    'to-read',
    'currently-reading',
    'read',
    'favorites',
    'kindle',
    'audiobook',
    'ebook',
    'library',
    'owned',
    'default',
    'my-books',
    'wishlist',
    'audio'
]

# Create a new column for cleaned genres
df['cleaned_genres'] = df['genres_list'].apply(lambda genres:
    [tag for tag in genres if tag not in non_genre_tags]
)

# Display some samples before and after cleaning
print("Sample of genres before and after cleaning:")
for i in range(5):
    print(f"Original: {df['genres_list'].iloc[i]}")
    print(f"Cleaned:  {df['cleaned_genres'].iloc[i]}\n")

# Update the all_shelf_tags to reflect the cleaned genres
all_cleaned_tags = defaultdict(int)
for tag in chain.from_iterable(df['cleaned_genres']):
    all_cleaned_tags[tag] += 1

print(f"Total unique cleaned genre tags found: {len(all_cleaned_tags)}")
print("Top 20 most frequent cleaned genre tags:")
for tag, count in sorted(all_cleaned_tags.items(), key=lambda item: item[1], reverse=True)[:20]:
    print(f"- {tag}: {count}")
```

```python
## Normalize Synonymous Tags

# Genre tags often have variations or synonyms (e.g., 'sci-fi' and 'science-fiction'). This step aims to consolidate these into a single, standardized tag to improve consistency and reduce dimensionality.
```

```python
from itertools import chain
from collections import defaultdict

# Display top 50 most frequent cleaned genre tags to help identify synonyms
all_cleaned_tags = defaultdict(int)
for tag in chain.from_iterable(df['cleaned_genres']):
    all_cleaned_tags[tag] += 1

print("Top 50 most frequent cleaned genre tags (to identify synonyms):")
for tag, count in sorted(all_cleaned_tags.items(), key=lambda item: item[1], reverse=True)[:50]:
    print(f"- {tag}: {count}")
```

Based on the frequent tags, here are some common synonyms I've identified that we should normalize. You can expand this list as needed for more comprehensive normalization.

For example:
*   `sci-fi` and `science-fiction` can be normalized to `science fiction`
*   `ya` and `young-adult` can be normalized to `young adult`
*   `fantasy-romance` and `paranormal-romance` might be normalized to `fantasy romance` and `paranormal romance` respectively, or even a broader `romance` category depending on the desired granularity.

```python
# Define a mapping for synonymous tags
synonym_map = {
    'sci-fi': 'science fiction',
    'science-fiction': 'science fiction',
    'ya': 'young adult',
    'young-adult': 'young adult',
    'romance-fantasy': 'fantasy romance',
    'fantasy-romance': 'fantasy romance',
    'paranormal-romance': 'paranormal romance',
    'historical-fiction': 'historical fiction',
    'mystery-thriller': 'mystery thriller',
    'thriller-mystery': 'mystery thriller',
    'graphic-novels': 'graphic novel',
    'graphic-novel': 'graphic novel',
    'comics': 'graphic novel',
    'middle-grade': 'middle grade',
    'urban-fantasy': 'urban fantasy',
    'contemporary-romance': 'contemporary romance',
    'dark-fantasy': 'dark fantasy',
    'high-fantasy': 'high fantasy',
    'epic-fantasy': 'epic fantasy',
    'magical-realism': 'magical realism',
    'dystopian': 'dystopia',
    'post-apocalyptic': 'post apocalyptic',
    'lgbt': 'lgbtq',
    'glbt': 'lgbtq',
    'lgbtqia': 'lgbtq'
}

def normalize_tags(genre_list, mapping):
    normalized = []
    for tag in genre_list:
        # Convert tag to lowercase for case-insensitive matching
        lower_tag = tag.lower()
        # Apply mapping, default to original tag if no synonym found
        normalized_tag = mapping.get(lower_tag, lower_tag)
        normalized.append(normalized_tag)
    return normalized

# Apply the normalization to the cleaned_genres column
df['normalized_genres'] = df['cleaned_genres'].apply(lambda x: normalize_tags(x, synonym_map))

print("Sample of cleaned genres and normalized genres:")
for i in range(5):
    print(f"Cleaned:    {df['cleaned_genres'].iloc[i]}")
    print(f"Normalized: {df['normalized_genres'].iloc[i]}\n")
```

```python
from itertools import chain
from collections import defaultdict

# Recalculate and display the top 20 most frequent normalized genre tags
all_normalized_tags = defaultdict(int)
for tag in chain.from_iterable(df['normalized_genres']):
    all_normalized_tags[tag] += 1

print(f"Total unique normalized genre tags found: {len(all_normalized_tags)}")
print("Top 20 most frequent normalized genre tags:")
for tag, count in sorted(all_normalized_tags.items(), key=lambda item: item[1], reverse=True)[:20]:
    print(f"- {tag}: {count}")
```

## Map to Canonical Genres

After normalizing synonymous tags, the next step is to map the refined genre tags to a predefined set of canonical (standardized) genres. This further reduces dimensionality and ensures consistency across the dataset, making the genre data more manageable for analysis or machine learning tasks.

```python
import collections
# Define a mapping from normalized tags to broader, canonical genres (or multiple canonical genres)
# Keys are lowercased normalized tags, values are lists of canonical genre strings (Title Cased)
# This map is designed to be less aggressive in collapsing genres.
canonical_genre_map = {
    'fiction': ['Fiction'],
    'fantasy': ['Fantasy'],
    'science fiction': ['Science Fiction'],
    'young adult': ['Young Adult'],
    'romance': ['Romance'],
    'mystery thriller': ['Mystery & Thriller'],
    'historical fiction': ['Historical Fiction'],
    'graphic novel': ['Graphic Novel'],
    'horror': ['Horror'],
    'biography': ['Biography & Memoir'],
    'memoir': ['Biography & Memoir'],
    'nonfiction': ['Non-Fiction'],
    'contemporary': ['Contemporary'],
    'adventure': ['Adventure'],

    # Split out specific sub-genres or map to multiple broader categories
    'dystopia': ['Science Fiction', 'Dystopia'],
    'urban fantasy': ['Fantasy', 'Urban Fantasy'],
    'high fantasy': ['Fantasy'],
    'epic fantasy': ['Fantasy'],
    'paranormal romance': ['Romance', 'Paranormal Romance'],
    'fantasy romance': ['Fantasy', 'Romance'],
    'lgbtq': ['LGBTQ+'],
    'historical romance': ['Historical Fiction', 'Romance'],
    'thriller': ['Mystery & Thriller'],
    'mystery': ['Mystery & Thriller'],
    'crime': ['Mystery & Thriller'],
    'literary fiction': ['Fiction', 'Literary Fiction'],
    'middle grade': ['Middle Grade'],
    'classic': ['Classics'],
    'classics': ['Classics'],
    'adult fiction': ['Fiction'],
    'magical realism': ['Fantasy', 'Magical Realism'],

    # Additional common genre tags that should be canonical
    'superhero': ['Superhero'],
    'action': ['Action'],
    'magic': ['Fantasy'],
    'western': ['Western'],
    'paranormal': ['Paranormal'],
    'erotica': ['Erotica'],
    'dark fantasy': ['Fantasy', 'Dark Fantasy'],
    'post apocalyptic': ['Science Fiction', 'Post-Apocalyptic'],
    'steampunk': ['Science Fiction', 'Fantasy', 'Steampunk'],

    # Ensure other variants of LGBTQ+ are covered if they didn't get normalized before
    'lgbt': ['LGBTQ+'],
    'glbt': ['LGBTQ+'],
    'lgbtqia': ['LGBTQ+'],
    'teen': ['Young Adult'],
}

def map_to_canonical_genres(normalized_genre_list, mapping):
    canonical = set()
    for tag in normalized_genre_list:
        mapped_genres = mapping.get(tag, None)
        if mapped_genres:
            canonical.update(mapped_genres)
        # If a normalized tag is not explicitly mapped, it is excluded from canonical.
        # This keeps the canonical list strictly to defined broader categories.
    return list(canonical) if canonical else []

# Apply the mapping to create a new 'canonical_genres' column
df['canonical_genres'] = df['normalized_genres'].apply(lambda x: map_to_canonical_genres(x, canonical_genre_map))

print("Sample of normalized genres and refined canonical genres:")
for i in range(5):
    print(f"Normalized: {df['normalized_genres'].iloc[i]}")
    print(f"Canonical:  {df['canonical_genres'].iloc[i]}\n")
```

```python
from itertools import chain
from collections import defaultdict

# Recalculate and display the top 20 most frequent canonical genre tags
all_canonical_tags = defaultdict(int)
for tag in chain.from_iterable(df['canonical_genres']):
    all_canonical_tags[tag] += 1

print(f"Total unique canonical genre tags found: {len(all_canonical_tags)}")
print("Top 20 most frequent canonical genre tags:")
for tag, count in sorted(all_canonical_tags.items(), key=lambda item: item[1], reverse=True)[:20]:
    print(f"- {tag}: {count}")
```

```python
import re
import pandas as pd
df=pd.read_csv('final_cleaned_data_with_target.csv')
# 1. Parse IDs
# The 'author_ids' column appears to contain string representations of lists.
# We'll parse these into actual Python lists using regex, similar to genre parsing.

print(f"Original type of 'author_ids' column: {df['author_ids'].dtype}")
print(f"Sample of original 'author_ids' data:\n{df['author_ids'].head()}\n")

def parse_author_ids(ids_str):
    if pd.isna(ids_str) or not isinstance(ids_str, str):
        return []
    # Use regex to find all numbers (author IDs) within the string, treating them as strings initially
    # Example format: "['12345' '67890']" or similar, we want to extract '12345' and '67890'
    author_tags = re.findall(r"'([^']*)'", ids_str)
    # Ensure the extracted IDs are consistent, e.g., convert to int if they represent numerical IDs
    return [int(tag) if tag.isdigit() else tag for tag in author_tags]

df['author_id_list'] = df['author_ids'].apply(parse_author_ids)

print(f"New 'author_id_list' column created. Sample of parsed 'author_id_list' data:\n{df['author_id_list'].head()}\n")
print(f"Type of 'author_id_list' column: {df['author_id_list'].dtype}\n")

# 2. Handle multiple authors (already handled by creating 'author_id_list' as a list)

# 3. Select a consistent primary author representation
# For simplicity, we'll take the first author ID from the list as the primary author.
# If the list is empty, the primary_author_id will be NaN.
df['primary_author_id'] = df['author_id_list'].apply(lambda x: x[0] if x else None)

print(f"New 'primary_author_id' column created. Sample of data:\n{df['primary_author_id'].head()}\n")

# 4. Validate author IDs / missing values
# Check for books where no author IDs could be parsed or primary author is missing.

books_no_author_ids = df[df['author_id_list'].apply(len) == 0]
print(f"Number of books with no parsed author IDs: {len(books_no_author_ids)}")

books_missing_primary_author = df[df['primary_author_id'].isnull()]
print(f"Number of books with a missing primary author ID: {len(books_missing_primary_author)}")

print("\nValue counts for primary_author_id (top 10):")
display(df['primary_author_id'].value_counts().head(10))
```
